```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

## 2 units with astropy

```python
df = pd.read_csv("solar_system.csv")
```

```python
df_tp = df.set_index("Attribute").T
df_tp.index.name = "Planet"
df_tp.reset_index(inplace=True)
df_tp.columns.name = None
```

```python
print(df.shape)
print(df_tp.shape)

# there were 20 rows and 11 columns before transposing, and 10 rows and 21 after transposing.

# c
# set_index takes the original dataframe and moves the data from the 'attribute'
# column as a new row index. 
# index.name names the new index
# reset_index moves the named index back into a regular column
# columns.name = None removes "name" from the columns axis

# d 
# the new shape is (10, 21) as it moves the "name" from the column axis
# so that all the data would be where it needs to be,

# e
print(df_tp.columns)

# f
# Mass, Diameter, Density, Gravity, Escape Velocity, Rotation Period, Perihelion,
# Aphelion, Orbital Period, Orbital Velocity, Orbital Inclination, Orbital Eccentricity,
# Obliquity to Orbit, Mean Temperature, Surface Pressure have units (15).
# Planet, Number of Moons, Ring System, and Global Magnetic Field do not (4)
```

```python
# investigate data types

df_tp.dtypes
```

```python

for col in df_tp.columns:
    type_counts = df_tp[col].apply(type).value_counts()
    print(type_counts)


# a: all of the attributes are strings, and not actual values.
# b: this makes it harder to work with calculations and applying numerical functions
```

```python
# convert strings to numbers
for col in df_tp.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        df_tp[col] = pd.to_numeric(df_tp[col], errors='coerce')

print(df_tp.dtypes)
```

```python
# the tree different data types are object, floats, and integers.
# planet, global magnetic field, and ring system are all still objects
# because they have no numerical value.
```

```python
# apply astropy units

def attach_units(dataframe, old_col, unit_val, new_col=None):
    column_data = dataframe[old_col].values
    dataframe[old_col] = column_data * unit_val
    if new_col is not None:
        df.rename(columns={old_col: new_col}, inplace=True)
        print(f"Column '{old_col}' renamed to '{new_col}' and units attached.")
    else:
        print(f"Units attached to column '{old_col}'.")

u_factor = u.hour
attach_units(df_tp, 'Rotation Period (hours)', unit_val = u_factor, new_col='hours')
```
