```python
# 4.3 reducing data

# planet name
    # database column name: pl_name†
    # description of parameter: planet name most commonly used in erliterature
    # description in own words: the name that the common population associates
        # with the planet based on how it is framed in literature
# host name
    # hostname†
    # Stellar name most commonly used in the literature
    # The name that people refer the planet's host as based on common association
# discovery method
    # discoverymethod†	
    # Method by which the planet was first identified
    # how the planet was studied, and how its criteria was collected to
        # classify it as a planet.
# spectral type
    # st_spectype
    # Classification of the star based on their spectral characteristics
        # following the morgan-keenan system
    # how the star is classified based on standard features
# equilibrium temperature [K]
    # pl_eqt†
    # The equilibrium temperature of the planet as modeled by a black body 
        #heated only by its host star, or for directly imaged planets, the 
        #effective temperature of the planet required to match the measured 
        #luminosity if the planet were a black body
    # The average temperature that a planet wuold reach if the only heat it
        # got was from its host, and the only way it cooled down was by 
        # radiating that heat back into space
# planet mass or mass*sin(i) [Jupiter Mass]
    # pl_msinij
    # Minimum mass of a planet as measured by radial velocity, measured in units
        #of masses of Jupiter
    # How massive a planet, contextualized by the size of jupiter
```
