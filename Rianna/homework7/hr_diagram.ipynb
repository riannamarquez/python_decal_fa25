```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print(np.max(magnitude))
print(np.min(magnitude))

print(np.argmin(magnitude))
print(np.argmax(magnitude))

## the minimum value of the magnitude variable corresponds to the star of 1988th index in the dataset
## the maximum value of the magnitude variable corresponds to the 264th index star of the dataset
```

```python
print(np.max(temperature))
print(np.min(temperature))
print(np.argmin(temperature))
print(np.argmax(temperature))

## the min value of temperature corersponds to the 1923rd indexed star.
## the max value of temperature corresponds to the 2613th indexed star of the dataset.
```

```python
# 3.3 plot and hr diagram
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.scatter(log_temperature, log_magnitude, c="blue", marker="o")
plt.scatter(log_temperature[1988], log_magnitude[1988], c="black", marker="*", s=300)
plt.scatter(log_temperature[264], log_magnitude[264], c="purple", marker="*", s=300)
plt.scatter(log_temperature[2613], log_magnitude[2613], c="green", marker="*", s=300)
plt.scatter(log_temperature[1923], log_magnitude[1923], c="orange", marker="*", s=300)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

#color = np.full(len(log_magnitude), "gray")
#color = np.full(len(log_temperature), "blue")
#color[1988] = "black"
#marker[1988] = "*"

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (logK)")
plt.ylabel("Magnitude")
plt.show()
```
